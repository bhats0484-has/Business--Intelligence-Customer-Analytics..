#**`Part 3: Customer RFM-PQD Feature Analysis`**

##**Objective:** Aggregating customer purchase patterns by Recency, Frequency, Sales (Monetary), Profit, Quantity, and Discounts to score overall buying behavior.
---

```python
import pandas as pd
import numpy as np

# 1. Reference date set up
snapshot_date = df['Order Date'].max() + pd.Timedelta(days=1)

# 2. Aggregate extended RFM + Profit, Discount & Quantity metrics
rfm_extended = df.groupby('Customer ID').agg({
    'Order Date': lambda x: (snapshot_date - x.max()).days, # Recency
    'Order ID': 'nunique',                                  # Frequency
    'Sales': 'sum',                                         # Monetary (Sales)
    'Quantity': 'sum',                                      # Total Units Purchased
    'Profit': 'sum',                                        # Net Profit
    'Discount': 'mean'                                      # Average Discount Rate
}).reset_index()

# Rename columns
rfm_extended.columns = [
    'Customer_ID', 'Recency', 'Frequency',
    'Monetary_Sales', 'Total_Quantity', 'Net_Profit', 'Avg_Discount'
]

# 3. Quantile Scoring (5-Tier Percentile Rank, 1-5 Scale)
# Recency (Reversed: Lower days = Higher rank)
rfm_extended['R_Score'] = pd.qcut(rfm_extended['Recency'], q=5, labels=[5, 4, 3, 2, 1])

# Frequency (Rank method handles repeated low values)
rfm_extended['F_Score'] = pd.qcut(rfm_extended['Frequency'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])

# Monetary (Higher sales = Higher rank)
rfm_extended['M_Score'] = pd.qcut(rfm_extended['Monetary_Sales'], q=5, labels=[1, 2, 3, 4, 5])

# Quantity (Higher order volume = Higher rank)
rfm_extended['Q_Score'] = pd.qcut(rfm_extended['Total_Quantity'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])

# 4. Composite RFM Segment & Score
rfm_extended['RFM_Segment'] = (
    rfm_extended['R_Score'].astype(str) +
    rfm_extended['F_Score'].astype(str) +
    rfm_extended['M_Score'].astype(str)
)

rfm_extended['Composite_RFM_Score'] = (
    rfm_extended['R_Score'].astype(int) +
    rfm_extended['F_Score'].astype(int) +
    rfm_extended['M_Score'].astype(int)
)

rfm_extended.head(10)
```

```python
# Custom function to map RFM Scores into Business Segments
def segment_customer(df):
    r = int(df['R_Score'])
    f = int(df['F_Score'])
    m = int(df['M_Score'])

    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'
    elif r >= 3 and f >= 3 and m >= 3:
        return 'Loyal Customers'
    elif r >= 4 and f <= 2:
        return 'Recent Buyers / Promising'
    elif r <= 2 and f >= 4:
        return 'At Risk (High Vol / Churn Risk)'
    elif r <= 2 and f <= 2 and m <= 2:
        return 'Lost Customers'
    else:
        return 'Needs Attention'

# Apply Segment Mapping
rfm_extended['Customer_Persona'] = rfm_extended.apply(segment_customer, axis=1)

# Segment Count Summary
rfm_extended['Customer_Persona'].value_counts().reset_index()
```

##**`Part 4: Pre-PCA Skewness Mitigation & Standardization`**

> **Technical Objective:** Log-transforming continuous metrics (`np.log1p`) to eliminate right-skewness and achieve normal distribution symmetry, followed by `StandardScaler` prior to PCA execution.
---

```python
# Skewness Mitigation: Yeo-Johnson Power Transformation (Handles Negatives & Extreme Skewness)
from sklearn.preprocessing import PowerTransformer

#selecting the continous features.
features = ['Recency', 'Frequency', 'Monetary_Sales', 'Total_Quantity', 'Net_Profit']
X = rfm_extended[features].copy()

# 2. Yeo-Johnson Transformer (Handles zero and negative values automatically)
pt = PowerTransformer(method='yeo-johnson')
X_log_data = pt.fit_transform(X)

#  Convert into the Dataframe.
X_log = pd.DataFrame(X_log_data, columns=features)

# 3. Diagnostic Skewness Comparison Table
skew_df = pd.DataFrame({
    'Before_Transform': X.skew(),
    'After_Transform': X_log.skew()
}).round(2)

print("Skewness Diagnostics Check")
print(skew_df)
```

```python
# Feature Standardization using StandardScaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_log)

# Scaling check
print("Scaled Matrix Shape:", X_scaled.shape)
```

```python
# PCA Execution & Variance Loss Verification
from sklearn.decomposition import PCA

# 1. Initialize PCA to retain 90% of total variance
pca = PCA(n_components=0.90)
X_pca = pca.fit_transform(X_scaled)

# 2. Check Individual & Cumulative Variance
var_explained = pca.explained_variance_ratio_ * 100
total_var = np.sum(var_explained)

print(f"Original Features Count : {X_scaled.shape[1]}")
print(f"Retained PCA Components : {X_pca.shape[1]}")
print(f"Total Information Kept  : {total_var:.2f}%")
print(f"Data Loss (Negligible)  : {100 - total_var:.2f}%")
print("-" * 40)

# Component-wise breakup
for i, var in enumerate(var_explained, 1):
    print(f"PC{i} Explains : {var:.2f}% Variance")
```

```python
# Create Reduced PCA DataFrame & Feature Loadings Matrix

# 1. Converting Reduced PCA Matrix into clean DataFrame
pc_colnames = [f'PC{i+1}' for i in range(X_pca.shape[1])]
X_pca_df = pd.DataFrame(X_pca, columns=pc_colnames, index=rfm.index)

# 2. Feature Loadings (Weights) Matrix Extract karna
loadings = pd.DataFrame(
    pca.components_.T,
    columns=pc_colnames,
    index=features
).round(3)

print(" Reduced PCA Dataset Sample (First 3 Rows)")
print(X_pca_df.head(3))

print("\nPCA Feature Loadings Matrix (Which Feature Controls Which PC)")
print(loadings)
```

#**`Part 5: K-Means Clustering - Optimal $k$ Selection`**

Evaluating $k \in [2, 8]$ on reduced PCA features (`X_pca_df`) using **Elbow Method (WCSS)** and **Silhouette Scores** to identify optimal customer segments.

```python
# Cell 1: Calculate WCSS and Silhouette Scores for k = 2 to 8

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

wcss = []
silhouette_scores = []
k_range = range(2, 9)

# Loop to evaluate KMeans for each k
for k in k_range:
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X_pca_df)

    wcss.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_pca_df, kmeans.labels_))

# Plotting Elbow Curve and Silhouette Scores
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4), dpi=100)

ax1.plot(k_range, wcss, 'bo-', linewidth=2, markersize=6)
ax1.set_title('Elbow Method (WCSS)', fontweight='bold')
ax1.set_xlabel('Number of Clusters (k)')
ax1.set_ylabel('Inertia (WCSS)')
ax1.grid(True, linestyle='--', alpha=0.5)

ax2.plot(k_range, silhouette_scores, 'ro-', linewidth=2, markersize=6)
ax2.set_title('Silhouette Score Analysis', fontweight='bold')
ax2.set_xlabel('Number of Clusters (k)')
ax2.set_ylabel('Silhouette Score')
ax2.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# Print Scores
for k, score in zip(k_range, silhouette_scores):
    print(f"Clusters (k={k}) -> Silhouette Score: {score:.4f}")
```

```python
# @title Train Final K-Means Model and Map Cluster Labels to PCA Data
# Cell 2: Train Final K-Means Model and Map Cluster Labels to PCA Data

# Set optimal k (e.g., k=4 based on Elbow & Silhouette analysis)
optimal_k = 4

kmeans_final = KMeans(n_clusters=optimal_k, init='k-means++', random_state=42, n_init=10)
cluster_labels = kmeans_final.fit_predict(X_pca_df)

# Attach Cluster Labels to both Reduced PCA DataFrame and Original RFM DataFrame
X_pca_df['Cluster'] = cluster_labels
rfm['Cluster'] = cluster_labels

# 1. Clear View of Reduced PCA Data Frame with Cluster Labels
print("Reduced PCA Data Frame with Cluster Labels (First 5 Rows)")
print(X_pca_df.head())

print("\nCluster Count Distribution")
print(X_pca_df['Cluster'].value_counts().sort_index().to_frame(name='Total Customers'))
```

```python
# @title 2D Visual Scatter Plot on PCA Space

# Cell 3: Clear 2D Visual Scatter Plot on PCA Space

import seaborn as sns

plt.figure(figsize=(9, 6), dpi=100)

# Scatter plot of PC1 vs PC2 colored by Cluster label
sns.scatterplot(
    x='PC1', y='PC2', hue='Cluster', palette='tab10',
    data=X_pca_df, s=70, alpha=0.85
)

# Plotting Centroids on PC1 & PC2 axes
centroids = kmeans_final.cluster_centers_
plt.scatter(
    centroids[:, 0], centroids[:, 1],
    s=250, c='black', marker='X', label='Cluster Centroids'
)

plt.title('K-Means Customer Segments in PCA Space (PC1 vs PC2)', fontweight='bold', fontsize=12)
plt.xlabel('PC1 (Customer Activity & Volume)')
plt.ylabel('PC2 (Profitability)')
plt.legend(title='Cluster Segment', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

#**`Phase 2: Customer Lifetime Value (CLV) Analysis`**
**Objective:** Quantify the financial value and lifetime profit contribution of each of the 4 K-Means RFM clusters.

**Key Metrics Computed:**
* **Average Order Value (AOV):** Total Revenue ÷ Total Order Frequency.
* **Annual Customer Value:** AOV × Annual Order Frequency.
* **4-Year Estimated CLV:** Total expected monetary spend over a 48-month retention window.
* **4-Year Profit Contribution:** Total net profit margin generated per customer segment.


```python
# @title Customer Lifetime Value (CLV) Analysis by Segments.

# Cell 6: Customer Lifetime Value (CLV) Analysis by Segments.

# 1. Aggregate RFM and Profit Metrics by Customer Segment
# Using rfm_extended which contains the full set of metrics
clv_summary = rfm_extended.groupby('Cluster').agg(
    Customer_Count=('Monetary_Sales', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary_Sales', 'mean'),
    Avg_Net_Profit=('Net_Profit', 'mean')
).reset_index()

# 2. Compute Average Order Value (AOV)
clv_summary['AOV'] = clv_summary['Avg_Monetary'] / clv_summary['Avg_Frequency']

# 3. Compute Customer Value (CV) per Year
clv_summary['Annual_Customer_Value'] = clv_summary['AOV'] * clv_summary['Avg_Frequency']

# 4. Estimate 2-Year Historical CLV & Expected Profit Contribution
clv_summary['Est_4Yr_CLV'] = clv_summary['Annual_Customer_Value'] * 4
clv_summary['Est_4Yr_Profit'] = clv_summary['Avg_Net_Profit'] * 4

# 5. Format Columns for Clear Presentation
clv_summary['AOV_fmt'] = clv_summary['AOV'].map('${:,.2f}'.format)
clv_summary['Avg_Monetary_fmt'] = clv_summary['Avg_Monetary'].map('${:,.2f}'.format)
clv_summary['Est_4Yr_CLV_fmt'] = clv_summary['Est_4Yr_CLV'].map('${:,.2f}'.format)
clv_summary['Est_4Yr_Profit_fmt'] = clv_summary['Est_4Yr_Profit'].map('${:,.2f}'.format)

# Display Final Table (Using Cluster as the identifier)
print("SEGMENT-WISE CUSTOMER LIFETIME VALUE (CLV) SUMMARY")
clv_summary[['Cluster', 'Customer_Count', 'AOV_fmt', 'Avg_Monetary_fmt', 'Est_4Yr_CLV_fmt', 'Est_4Yr_Profit_fmt']]
```

```python
# @title CLV and Profit Comparisons
# Cell 7: Plot CLV and Profit Comparisons

# 1. Prepare data for plotting by ensuring numeric types and segment names
clv_summary_plot = clv_summary.copy()
# Use 'Cluster' as the name since 'Segment_Name' isn't defined
clv_summary_plot['Segment_Name'] = clv_summary_plot['Cluster'].apply(lambda x: f'Cluster {x}')

# 2. Create subplots for CLV and Net Profit
fig, axes = plt.subplots(1, 2, figsize=(14, 5), dpi=100)

# Graph 1: 4-Year Estimated CLV by Segment
sns.barplot(
    x='Est_4Yr_CLV',
    y='Segment_Name',
    data=clv_summary_plot,
    ax=axes[0],
    palette='Blues_r',
    hue='Segment_Name',
    legend=False
)
axes[0].set_title('Estimated 4-Year CLV ($) by Segment', fontweight='bold', fontsize=12)
axes[0].set_xlabel('Estimated CLV ($)')
axes[0].set_ylabel('Customer Segment')

# Graph 2: 4-Year Net Profit Contribution by Segment
sns.barplot(
    x='Est_4Yr_Profit',
    y='Segment_Name',
    data=clv_summary_plot,
    ax=axes[1],
    palette='Greens_r',
    hue='Segment_Name',
    legend=False
)
axes[1].set_title('Estimated 4-Year Net Profit ($) by Segment', fontweight='bold', fontsize=12)
axes[1].set_xlabel('Estimated Net Profit ($)')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

#**`Phase 3: Cohort Retention Analysis (Monthly Matrix)`**
**Objective:** Track customer retention rates over time by grouping customers into acquisition cohorts based on their first purchase month.

**Key Concepts:**
* **Cohort Group:** Month of first purchase.
* **Cohort Index:** Number of months elapsed since the first purchase ($0, 1, 2, \dots, n$).
* **Retention Rate:** Percentage of the original cohort returning in subsequent months.


```python
# @title Cohort Retention Analysis and Heatmap Plot
# Cell 8: Cohort Retention Analysis and Heatmap Plot

# 1. Ensure OrderDate is in datetime format (Using your original order dataset)
df['OrderDate'] = pd.to_datetime(df['Order Date'])

# 2. Extract Year-Month for transaction date
df['OrderMonth'] = df['OrderDate'].dt.to_period('M')

# 3. Determine First Purchase Month for each customer (Cohort Group)
df['CohortGroup'] = df.groupby('Customer ID')['OrderDate'].transform('min').dt.to_period('M')

# 4. Calculate Cohort Index (Months elapsed since first purchase)
df['CohortIndex'] = (df['OrderMonth'].dt.year - df['CohortGroup'].dt.year) * 12 + \
                    (df['OrderMonth'].dt.month - df['CohortGroup'].dt.month)

# 5. Build Cohort Matrix (Count of Unique Customers)
cohort_data = df.groupby(['CohortGroup', 'CohortIndex'])['Customer ID'].nunique().reset_index()
cohort_matrix = cohort_data.pivot(index='CohortGroup', columns='CohortIndex', values='Customer ID')

# 6. Calculate Retention Rate Percentage (%)
cohort_size = cohort_matrix.iloc[:, 0]
retention_matrix = cohort_matrix.divide(cohort_size, axis=0) * 100

# 7. Plot Cohort Retention Heatmap
# Updated Cell 8 Plotting Code (Clean & Readable Version)
plt.figure(figsize=(20, 10), dpi=150) # Sizing increase ki hai
plt.title('Monthly Customer Retention Rate (%) Heatmap', fontsize=16, fontweight='bold', pad=20)

sns.heatmap(
    retention_matrix,
    annot=True,
    fmt='.0f',          # Decimal points hata diye (.0f) taaki clean integers dikhein
    annot_kws={'size': 7}, # Numbers ka font size chhota kar diya
    cmap='Blues',
    cbar_kws={'label': 'Retention Rate (%)'},
    linewidths=0.5      # Cells ke beech clean grid border ke liye
)

plt.xlabel('Months Elapsed (Cohort Index)', fontweight='bold', fontsize=12)
plt.ylabel('Acquisition Cohort (First Purchase Month)', fontweight='bold', fontsize=12)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

#**`Phase 4: Time-Series Revenue Forecasting (ARIMA Model)`**
**Objective:** Forecast upcoming monthly sales trends to help inventory management align with predicted customer demand across clusters.

**Key Steps:**
1. Resample daily sales data to Monthly Total Sales (`MS`).
2. Train ARIMA (AutoRegressive Integrated Moving Average) model.
3. Forecast revenue for the next 6 to 12 months with confidence intervals.


```python
# @title Seasonal Decomposition
# Cell 9A: Basic Time-Series Exploration & Seasonal Decomposition
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# 1. Monthly Aggregate Sales
ts_data = df.set_index('OrderDate')['Sales'].resample('MS').sum()

# 2. Seasonal Decomposition (Trend + Seasonality + Noise)
decomposition = seasonal_decompose(ts_data, model='additive')

# 3. Plotting Baseline Components
fig, axes = plt.subplots(4, 1, figsize=(12, 8), sharex=True, dpi=100)

axes[0].plot(ts_data, label='Original Sales', color='#1f77b4')
axes[0].set_title('Historical Monthly Sales Baseline', fontweight='bold')
axes[0].legend(loc='upper left')

axes[1].plot(decomposition.trend, label='Trend', color='#ff7f0e')
axes[1].set_title('Overall Growth Trend')
axes[1].legend(loc='upper left')

axes[2].plot(decomposition.seasonal, label='Seasonality', color='#2ca02c')
axes[2].set_title('Yearly Seasonal Patterns')
axes[2].legend(loc='upper left')

axes[3].plot(decomposition.resid, label='Residuals (Noise)', color='#d62728')
axes[3].set_title('Random Noise')
axes[3].legend(loc='upper left')

plt.tight_layout()
plt.show()
```
